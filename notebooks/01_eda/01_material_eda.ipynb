# Material EDA
Reproducible exploration of DS1 with physical-quality gates.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
DATA=Path('../../data/raw')
df=pd.read_csv(DATA/'DS1_material_properties_5500.csv')
df.info(); df.describe(include='all').T
```

```python
numeric=df.select_dtypes('number')
axes=numeric.hist(figsize=(18,14),bins=30)
plt.suptitle('Material property distributions'); plt.tight_layout()
```

```python
fig,axes=plt.subplots(2,2,figsize=(14,10))
for ax,(x,y) in zip(axes.flat,[('bulk_modulus_GPa','shear_modulus_GPa'),('formation_energy_per_atom_eV','band_gap_eV'),('density_g_cm3','melting_point_K'),('spacegroup_number','energy_above_hull_eV')]):
    ax.scatter(df[x],df[y],s=5,alpha=.25); ax.set(xlabel=x,ylabel=y)
plt.tight_layout()
```
