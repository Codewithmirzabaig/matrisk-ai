# Financial and infrastructure EDA
Point-in-time exploration of DS2–DS6.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
DATA=Path('../../data/raw')
prices=pd.read_csv(DATA/'DS2_commodity_prices_10yr.csv',parse_dates=['date'])
fusion=pd.read_csv(DATA/'DS4_crossdomain_features_daily.csv',parse_dates=['date'])
bridges=pd.read_csv(DATA/'DS3_infrastructure_bridges_5000.csv')
```

```python
fig,axes=plt.subplots(4,2,figsize=(16,16))
for ax,(name,part) in zip(axes.flat,prices.groupby('commodity')):
    part.plot(x='date',y='close',ax=ax,title=name,legend=False)
plt.tight_layout()
```

```python
merged=prices.merge(fusion,on=['date','commodity'],validate='one_to_one')
fig,axes=plt.subplots(2,3,figsize=(16,9))
for ax,column in zip(axes.flat,['mqi','supply_disruption_prob','substitution_elasticity','green_premium_per_kg','herfindahl_index','mqi_21d_trend']):
    ax.scatter(merged[column],merged['return_21d'],s=3,alpha=.15); ax.set(xlabel=column,ylabel='21d return')
plt.tight_layout()
```
